#### <font face="Verdana" color="green" size="+2">**1.1. (2 points)**

What challenges arise when the dimensionality of data increases, and how do these challenges impact the performance of machine learning models?</font>

When the dimensionality of data increases, challenges arise due to the "curse of dimensionality." This refers to the exponential growth of the data space as dimensions increase, which can lead to sparsity and make it harder to identify meaningful patterns. Key challenges include:

* Overfitting: High-dimensional data can cause models to fit noise instead of the true signal, reducing generalization to new data.
* Increased computational cost: Higher dimensions require more resources for processing and training, leading to slower performance.
* Difficulty in visualization: Interpreting and understanding high-dimensional data becomes impractical.
* Feature irrelevance: Many dimensions may be redundant or irrelevant, adding noise rather than value to the model.
* Training problems: when the dimension increases (and sample size stays the same) the distribution became more sparse, making it difficult for different algorithms to perform as they are expected. For example, since the 'Distance' between points become larger, the clustring algorithms have more trouble to find the best clusters.

These challenges often degrade model performance by increasing error rates and reducing efficiency, necessitating techniques like dimensionality reduction (e.g., PCA) or feature selection to mitigate these issues.

#### <font face="Verdana" color="green" size="+2">**1.2. (2 points)**

Once data has undergone dimensionality reduction, is it possible to recover the lost information? If so, what limitations might we face in doing so? </font>

Recovering lost information after dimensionality reduction is partially possible but comes with limitations. Techniques like inverse transformations (e.g., using the inverse of PCA or autoencoders) attempt to reconstruct the original data. However, the success of this process depends on the method used and the extent of the reduction.

* Information loss: Irrelevant or redundant features might be removed during dimensionality reduction, but critical nuances can also be lost, making exact reconstruction impossible.
* Assumptions of the model: Reconstruction assumes the reduced dimensions capture most of the variance or structure, which may not hold true for highly complex data.
* Approximation errors: Methods like PCA prioritize variance, not precision, leading to approximations in reconstruction rather than true recovery.

#### <font face="Verdana" color="green" size="+2">**1.3. (2 points)**

In what situations is Principal Component Analysis (PCA) unsuitable for nonlinear data, and why? </font>

PCA is unsuitable for nonlinear data when the relationships between features are complex and not captured by linear correlations. This is because PCA relies on finding linear combinations of features (principal components) that maximize variance.

In cases where data has inherent nonlinear structures, like curves or clusters, PCA may fail to identify and preserve these patterns. This leads to inaccurate representations in the reduced dimensions, as PCA attempts to fit a linear model to nonlinear relationships, resulting in significant information loss and potentially misleading interpretations. For nonlinear data, consider methods like t-SNE or kernel PCA, which are better equipped to handle complex relationships.

Standard PCA is suitable for smaller datasets that fit in memory and when the computational cost is not a concern. It provides precise results by computing exact eigenvalues and eigenvectors but becomes inefficient for very large datasets.

Incremental PCA is ideal for large datasets that cannot fit into memory at once. It processes data in mini-batches, making it efficient for memory-constrained environments while providing results comparable to standard PCA. However, it may be slightly slower due to its iterative nature.

Randomized PCA is well-suited for large datasets where an approximate solution is sufficient. ***Elbow method:***

One way to determine the appropriate number of clusters for a k-means clustering model is by using the elbow method. It is based on inertia, which measures the sum of squared distances between each data point and its nearest centroid.
This involves plotting the within-cluster sum of squares (WCSS) against the number of clusters (k). As k increases, the WCSS decreases because the clusters become smaller. The "elbow" point on the graph, where the rate of decrease slows significantly, suggests the optimal number of clusters. This method is most effective when the data has clear, well-separated clusters, as the elbow is easier to identify.

alternatvie text


The WCSS for $k$ clusters is calculated as:

$$
\text{WCSS} = \sum_{i=1}^{k} \sum_{x \in C_i} \|x - \mu_i\|^2
$$

Where:
- $C_i$ is the $i$-th cluster.
- $x$ is a data point in cluster $C_i$.
- $\mu_i$ is the centroid of the $i$-th cluster.
- $\|\cdot\|$ denotes the Euclidean norm (distance).


**Silhouette Score**

Another approach is the silhouette score, which measures how well each point fits within its cluster compared to other clusters. The score ranges from -1 to 1, with higher values indicating better-defined clusters. By calculating the silhouette score for different k values, the k with the highest average score can be chosen. This method is useful when the clusters may overlap or are not well-separated, as it directly evaluates the quality of clustering.

The **Silhouette Score** measures how well each data point fits within its assigned cluster compared to how well it fits in other clusters. The silhouette score for a data point is calculated as:

$$
S = \frac{(b - a)}{\max(a, b)}
$$

Where:

- **a** is the mean intra-cluster distance (the average distance between a point and other points within the same cluster).

 It is given by:

$$
a(i) = \frac{1}{|C_i| - 1} \sum_{j \in C_i, j \neq i} \|x_i - x_j\|
$$

Where:
- $C_i$ is the cluster to which point $i$ belongs.
- $|C_i|$ is the number of points in cluster $C_i$.
- $x_i$ and $x_j$ are data points.
- $\|\cdot\|$ denotes the Euclidean distance (or another distance metric).


**b** is the mean nearest-cluster distance (the average distance between a point and the points in the nearest cluster). It is given by:

$$
b(i) = \min_{C_k \neq C_i} \frac{1}{|C_k|} \sum_{j \in C_k} \|x_i - x_j\|
$$

Where:
- $C_k$ is a cluster different from $C_i$.
- $|C_k|$ is the number of points in cluster $C_k$.
- The minimum is taken over all clusters $C_k$ that are not $C_i$.

Intuition:
- $a(i)$ measures how compact the cluster $C_i$ is (lower values are better).
- $b(i)$ measures how far $i$ is from the nearest other cluster (higher values are better).


Silhouette Score Range:
The score ranges from -1 to 1.

- A value close to **1** indicates that the point is well clustered.
- A value close to **0** means that the point is on or very close to the decision boundary between clusters.
- A negative value indicates that the point might have been assigned to the wrong cluster.

The optimal **K** is the one that maximizes the silhouette score, indicating that the clusters are well-separated and compact.

Label propagation is a semi-supervised learning algorithm that propagates labels from a small set of labeled data to unlabeled data points based on their similarity in a graph. #### <font face="Verdana" color="green" size="+2">**1.8. (2 points)**

Identify two clustering algorithms suited for large-scale datasets, and two that are designed to detect high-density regions. How do these approaches differ in clustering methodology? </font>

Two clustering algorithms suited for large-scale datasets are K-means and Mini-Batch K-means. K-means is efficient for large datasets because it uses centroid-based clustering and updates centroids iteratively. Mini-Batch K-means is a variant that uses small batches of data at each iteration, speeding up the process and reducing memory usage, making it more suitable for even larger datasets.

Two algorithms designed to detect high-density regions are DBSCAN (Density-Based Spatial Clustering of Applications with Noise) and HDBSCAN (Hierarchical DBSCAN). These methods identify clusters based on areas with high data density, allowing them to find clusters of arbitrary shape and detect outliers, unlike centroid-based methods.


* DBSCAN Algorithm

DBSCAN (Density-Based Spatial Clustering of Applications with Noise) is a clustering algorithm that groups data points based on their density. Here's how it works:

1. **Core Points**: Points with at least a minimum number of neighbors (MinPts) within a specified radius (ε, epsilon) are identified as core points.
2. **Cluster Formation**: Starting from a core point, clusters are expanded by including all directly reachable points (within ε) and iteratively adding their neighbors that meet the density requirement.
3. **Border Points**: Points within ε of a core point but not meeting the MinPts requirement themselves are added to the cluster as border points.
4. **Noise**: Points that are neither core points nor reachable from any core point are classified as noise.



The main difference in clustering methodology is that K-means (and Mini-Batch K-means) partition data into spherical clusters based on distances to centroids, assuming that clusters are roughly equal in size. In contrast, DBSCAN and HDBSCAN focus on regions of high data density and do not require the number of clusters to be specified beforehand. These density-based methods are better at handling clusters of varying shapes and sizes and are robust to noise.

What distinguishes anomaly detection from novelty detection, and in which scenarios would each be more applicable? </font>

Anomaly detection and novelty detection both aim to identify data points that deviate from the norm, but they differ in their focus and application. Anomaly detection identifies data points that significantly differ from the rest of the data in a given dataset, typically using a model trained on historical data. It’s used in scenarios where the dataset contains a mix of normal and anomalous points, such as fraud detection or network intrusion.

On the other hand, novelty detection focuses on identifying new or previously unseen patterns that were not present in the training data. It’s applied when the model is exposed to new data that may contain previously unknown examples, such as in cases where the system is continually learning from new input (e.g., in adaptive systems or robotics).

#### <font face="Verdana" color="green" size="+2">**1.10. (2 points)**

Discuss two approaches to determine the optimal number of clusters in a Gaussian Mixture Model and the underlying assumptions of each approach. </font>

**AIC (Akaike Information Criterion)** and **BIC (Bayesian Information Criterion)** are both used for model selection by balancing model fit and complexity.

- **AIC** is calculated as:

  $$
  \text{AIC} = 2k - 2 \ln(L)
  $$

  where $ k $ is the number of parameters in the model, and $ L $ is the likelihood of the model. AIC penalizes the complexity of the model but tends to favor more parameters compared to BIC.

- **BIC** is calculated as:

  $$
  \text{BIC} = \ln(n)k - 2 \ln(L)
  $$

  where $ n $ is the number of data points, $ k $ is the number of parameters, and $ L $ is the likelihood of the model. BIC applies a stronger penalty for more parameters than AIC, making it more likely to select simpler models.

Both metrics aim to prevent overfitting by penalizing the number of parameters, but BIC typically prefers simpler models, especially with large sample sizes.

Analyze the result </font>

K-modes is a clustering algorithm designed for categorical data, similar to k-means but using modes instead of means. It begins by randomly selecting k initial cluster modes, representing the most frequent categories within each cluster. Each data point is then assigned to the cluster whose mode is most similar to it, measured by a dissimilarity function (often simple matching distance). This dissimilarity is calculated based on the number of mismatched attributes between the data point and the cluster mode. The goal is to minimize the total dissimilarity within each cluster, ensuring data points within a cluster share similar categorical values.

After the initial assignment, k-modes iteratively refines the clusters. The mode of each cluster is recalculated to reflect the most frequent categories among its assigned data points. Data points are then reassigned to clusters based on the updated modes. This iterative process of mode recalculation and data point reassignment continues until cluster assignments stabilize, indicating convergence. The final output consists of cluster assignments for each data point and the final modes for each cluster, offering a descriptive summary of the categorical features within those clusters.

*   **Analysis:** The bar chart displays the size of each cluster.  Clusters with a large number of data points suggest common characteristics among those data points.  The cluster centroids, printed from `kmode.cluster_centroids_`, provide insights into the defining features of each cluster.
