**Task:**

The European Society of Cardiology provides a SCORE scale on their website, which is used to estimate the 10-year risk of death from cardiovascular disease. Here is the scale

![score_cvd](https://s8.uupload.ir/files/score_cvd_swga.png)

Let’s analyze the upper-right section of the SCORE chart, which focuses on smoking men aged 60 to 65. The values in this chart represent upper bounds for different risk categories.

For instance, the value "9" in the lower-left corner and "47" in the upper-right corner indicate that, within this gender and age group, men with systolic blood pressure below 120 have a cardiovascular disease (CVD) risk five times lower than those with systolic pressure in the 160-180 range.

Now, let's calculate this risk ratio using our dataset.

**Clarifications:**

- **Age Calculation**: Create a new feature, `age_years`, by rounding the age to the nearest year. For this analysis, only include individuals aged 60 to 64, inclusive.

- **Cholesterol Conversion**: The cholesterol levels in our dataset differ from those in the figure. Use the following conversion for the `cholesterol` feature:
  - 4 mmol/L corresponds to category 1
  - 5-7 mmol/L corresponds to category 2
  - 8 mmol/L corresponds to category 3

Let’s proceed with the analysis based on these criteria.

````python
male_smokers = (data[(data['gender_label'] == 'Male') & (data['smoke'] == 1)])
data_60_64 = male_smokers[(male_smokers['age_years'] >= 60) & (data['age_years'] <= 64)]
````

````python
data_60_64.head(5)
````

#### **<font face="Courier New" color="red" size="+3">2.4. Cleaning data</font>**

It's clear that the data isn't perfect; it contains some `dirty`. This will become more evident as we start visualizing the data.

To clean up the dataset, filter out the following segments of patients, as these are considered erroneous data:

- Cases where diastolic pressure is higher than systolic pressure.
- Patients whose height is below the 2.5th percentile (use `pd.Series.quantile` to compute this value—refer to the documentation if you're unfamiliar with this function).
- Patients whose height is above the 97.5th percentile.
- Patients whose weight is below the 2.5th percentile.
- Patients whose weight is above the 97.5th percentile.

This isn't an exhaustive cleaning process, but it’s a good starting point for now.

````python
# we filter out (throw away) all rows that satisfies one of the above conditions. I do this by keeping the rest of the dataset.
filtered_dataset = data.iloc[np.where(data['ap_lo'] <= data['ap_hi'])]
filtered_dataset = filtered_dataset.iloc[np.where(filtered_dataset['weight'] >= pd.Series(data['weight']).quantile(0.025))]
filtered_dataset = filtered_dataset.iloc[np.where(filtered_dataset['weight'] <= pd.Series(data['weight']).quantile(0.975))]
filtered_dataset = filtered_dataset.iloc[np.where(filtered_dataset['height'] <= pd.Series(data['height']).quantile(0.975))]
filtered_dataset = filtered_dataset.iloc[np.where(filtered_dataset['height'] >= pd.Series(data['height']).quantile(0.025))]
filtered_dataset.head(5)
````

#### <font face="Verdana" color="green" size="+2">**3.1. (3 points)**

Explain the diffrence between <u>parameters</u> and <u>hyperparameters</u>. </font>

Parameters are variable that are internal of the model. They are learnable values and will change during the training process by our model in order to minimize the cost function. For instance, feature **Weights** are model parameters that are adjusted by model.

Hyperparameters are those variables that are set **before** the training process by **user**, and they have a huge impact on the performance of the model. They usually don't change during training process. A good example of a hyperparameter in a ML project is **Learning-Rate** . (Keep in mind that hyperparameters are not normally learnable, but, we can change them manually, or automaticcaly and programmaticly, during the learning process.)


#### <font face="Verdana" color="green" size="+2">**3.2. (3 points)**

Describe the diffrence between the <u>Validation Set</u> and the <u>Test Set</u></font>

In the begining, we split the data into 3 sets, training set, validation set, and test set.

***Test Set:*** We completely put away the test set during training phase. This set will be used only **once**, when the model is selected and completely tuned, and we want to calculate the ultimate model's generalization performance on unseen data to report.

***Validation Set:*** validation set is used to fine tune the hyperparameters of the model, or used for early-stopping purposes. When we have a bunch of different models ( different hyperparameters will also lead to diffrent models) we should be able to peak the best model among them. It's wrong to choose the best model based on the 'training loss',since it would lead to overfitting.  thus, we should have some kind of unseen dataset to evaluate the performance of each model on it. it's obvious that we cannot user test set for this purpose. So, we create kind of unseen dataset called validation set for this purpose.


#### <font face="Verdana" color="green" size="+2">**3.3. (3 points)**

Why is the <u>validation set</u> neccessary in hyperparameter optimization?</font>

In hyperparameter optimization, the aim is to find the best setting for hyperparameters which results in the best performance. We need to evaluate and compare model performances in order to select the best setting. but we cannot use training loss, As we probably risk overfitting the model. If we use training loss, we can adjust the parameters to have almost 0 loss! but it is almost overfitting, and will have a very poor result on the test set.

 Thus, we should evaluate the performance on unseen data. The validation set helps with an unbiased evaluation of performance on unseen data.

#### <font face="Verdana" color="green" size="+2">**3.4. (3 points)**

What are the diffrences between <u>Feature Selection</u> and <u>Feature Extraction</u> methods?</font>

**Feature Selection:** choose a subset of original features to have a lower-dimension feature space. We try to give each variable a score, or rank them based on some criteria, then select the k best features which has more impact on the target variable. Some of the methods are: forward-selection, backward-elimination, Ridge Regression, Lasso Regression, decision trees' feature importance method (based on gini-impurity), etc.


**Feature Extraction:** combine the original features to obtain new features. (transform original feature space into a new feature space, often with fewer dimension.) Some famous methods of feature extraction are : Principle Component Analysis (PCA) and its variations (Kernel PCA, Increamental PCA, ...) , Independent Component Analysis(ICA) t-SNE, UMAP, AutoEncoders. In the domain of Natural Language Processing, we have 2 other methods for feature extraction: Bag of Words (BoW), and TF-IDF.

1. Simply, drop all columns that have more than 15 percent null ratio. It's hard too fill missing values for these columns using other techniques like replacing with mean, or median.

````python
columns_to_drop = null_ratios[null_ratios > 15].index.tolist()
ames_housing_filtered = ames_housing.drop(columns=columns_to_drop)
print(f"Columns dropped: {columns_to_drop}")
print(f"Shape of filtered DataFrame: {ames_housing_filtered.shape}")
````

````python
#fill Nan values with median of their column (or mode for categorical columns) , if the total number of missing values in that column is less than 100

def fill_missing_values(df):

  for column in df.columns:
    if df[column].isnull().sum() < 100:
      if pd.api.types.is_numeric_dtype(df[column]):
        df[column].fillna(df[column].median(), inplace=True)
      else:
        df[column].fillna(df[column].mode()[0], inplace=True)
  return df


# Example usage:
# Assuming 'filtered_dataset' is your DataFrame
ames_housing_filtered = fill_missing_values(ames_housing_filtered)
````

````python
from sklearn.model_selection import RandomizedSearchCV

# Define the parameter grid for random search
param_grid = {
    'n_estimators': [25, 50, 100, 200],
    'max_depth': [ 10, 15, 20, 30],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4],
    'bootstrap': [True, False]
}

# Create a Random Forest model
rf_model = RandomForestRegressor(random_state=42)

# Create a RandomizedSearchCV object
random_search = RandomizedSearchCV(
    estimator=rf_model,
    param_distributions=param_grid,
    n_iter=10,  # Number of parameter settings that are sampled
    scoring='neg_mean_squared_error',
    cv=3,
    random_state=42,
    verbose = 10
    )

random_search.fit(X_train, y_train)

# Get the best model and its parameters
best_rf_model = random_search.best_estimator_
best_params = random_search.best_params_

print("Best hyperparameters:", best_params)

# Make predictions on the test set using the best model
y_pred_best = best_rf_model.predict(X_test)

# Evaluate the best model
rmse_best = np.sqrt(mean_squared_error(y_test, y_pred_best))
r2_best = r2_score(y_test, y_pred_best)

print("\nRandom Forest (Best):")
print("RMSE:", rmse_best)
print("R-squared:", r2_best)
````
